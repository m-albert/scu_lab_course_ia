# openCV tutorial and demo

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from iaf.plot import imshow
from iaf.notebook import init_style
init_style()
```

```python
# Open the sudoku image
img = cv2.imread('sudoku.png', 0)
```

```python
# Display the image
imshow(img, title="Original")
```

```python
# Run the laplacian and the horizontal and vertical sobel filters on the image
# Please notice that we set the data type of the filtered images to floating point 64-bit (double precision).
# The laplacian and sobel filters generate both positive and negative values, and if we kept the data type of the
# output the same as the data type of the input image (unsigned int 8 bit) all the negative values would be clipped to 0!
laplacian = cv2.Laplacian(img, cv2.CV_64F)
sobelx = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=5)
sobely = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=5)
```

```python
# Plot the laplacian
imshow(laplacian, title="Laplacian")
```

```python
# Plot the sobel x
imshow(sobelx, title = "Sobel X")
```

```python
# And finally the sobel y
imshow(sobely, title = "Sobel Y")
```
