1. Define variables that hold image metadata and print them (e.g. filename, width, height, and pixel_size).

```python
filename = "sample_image.tif"
width = 1024
height = 768
pixel_size = 6.5  # micrometers per pixel

print(f"Image: {filename}, Size: {width}x{height} pixels, Pixel spacing: {pixel_size} µm")
```

2. Given a pixel intensity value, print if it is above or below a threshold

```python
pixel_intensity = 120
threshold = 100

if pixel_intensity > threshold:
    print("Pixel is bright")
else:
    print("Pixel is dark")
```

3. Simulate iterating over a list of pixel intensities and count how many are above a threshold.

```python
pixels = [50, 120, 200, 80, 90, 150]
count_bright = 0
threshold = 100

for p in pixels:
    if p > threshold:
        count_bright += 1

print(f"Number of bright pixels: {count_bright}")
```

4. Write a function that takes a list of pixel intensities and returns the average intensity.

```python
def average_intensity(pixel_values):
    total = sum(pixel_values)
    count = len(pixel_values)
    return total / count

pixels = [50, 120, 200, 80, 90, 150]
avg = average_intensity(pixels)
print(f"Average pixel intensity: {avg}")
```
